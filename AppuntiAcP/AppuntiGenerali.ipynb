```python
# Prende una tupla di parametri 
def f(*args):
    print("Sono nella funzione f.")
    print(f"Valore di *args: {args}")
    for arg in args: print(arg)

# Prende un dizionario di valori, associando ad ogni valore una chiave
def g(**kwargs):
    print("\nSono nella funzione g.")
    print(f"Valore di **kwargs: {kwargs}")
    for chiave,valore in kwargs.items():
        print(f"{chiave} = {valore}")

def h(*args,**kwargs):
    print("\nSono nella funzione h.")
    for arg in args: print(arg)
    for chiave,valore in kwargs.items(): print(f"{chiave} = {valore}")

f(1,2,"ciao")
g(nome = "mario", cognome = "caterino")
# Posso usare la notazione posizionale e DOPO quella a keyword, ma non
# il contrario.
h(5,6,numero = "dieci")
```

```python
# Il blocco finally viene eseguito prima di un eventuale return
# Altre info: puoi decidere se salvare l'eccezione in una variabile

def f(a,b):

    # Se vuoi lanciare un eccezione una la parola chiave raise.
    # ESEMPIO: togli il commento se vuoi testare questa cosa
    # 
    # if a == 2: raise Exception("Stai inserendo un numero che non mi piace.")
    # 

    try: 
        print(a/b)
    except ZeroDivisionError as e:
        print(f"Errore divisione per zero!\n{e}")
        return -1
    except Exception:
        print("Errore eccezione generica!")
        return -2
    finally:
        print("Sono nel blocco finally")

a = 2
b = 0

ret = f(a,b)
print(ret)
```

```python
# Posso usare un "decoratore" con @qualcosa
# Esempio:

class Rabbit:
    # Variabile globale.
    tag = 1

    def __init__(self):
        Rabbit.tag += 1

    @staticmethod
    def get_tag():
        return Rabbit.tag

print(Rabbit.get_tag())
r1 = Rabbit()
r2 = Rabbit()
print(r1.get_tag())
print(r2.get_tag())
```

```python
# Per creare una classe astratta uso il modulo ABC -> Abstract Base Class
from abc import ABC,abstractclassmethod

# Ogni classe astratta è classe figlia di ABC
class AbstractClass(ABC):
    def __init__(self):
        pass
    
    # I metodi astratti sono definiti con i decoratori
    @abstractclassmethod
    def DoSomething(self):
        pass

# Se non voglio implementare il metodo in classi figlie posso lanciare una
# raise NotImplementedError()
class Abs2(AbstractClass):
    def __init__(self):
        pass

    def DoSomething(self):
        raise NotImplementedError("Non ho ancora implementato il metodo.")

# Non posso instanziare un oggetto che non ha definito un metodo
class NotAbs(AbstractClass):
    def __init__(self):
        pass
    
    def ciao():
        print("ciao")

a = NotAbs()
a.DoSomething()
b = Abs2()
b.DoSomething()
```

```python
from threading import local,Thread
# Permette di definire delle variabili locali all'interno di un thread
import logging
# Permette di scrivere dei log per alcuni parametri

logging.basicConfig(level=logging.DEBUG, # Severity
                    format='((%(threadname)-0s) %(message)s)') # Formato della stampa
# RIVEDI IL FORMATO

#Esempio di funzione che stampa un messaggio di log
def show(d):
    try:
        ret = d.val
    except AttributeError:
        logging.debug("Nessun valore")
    else:
        logging.debug(f"Valore: {ret}")

if __name__ == "__main__":
    d = local()
    d.val = 10
    show(d)

    t = Thread(target=show,args=(d,))
    t.start()

    t.join()
```

```python
# Per poter sfruttare al meglio il parallelismo uso la libreria di multiprocess
import multiprocessing as mp

def f(PipeConnection):
    PipeConnection.send(10)
    PipeConnection.close()

def g(PipeConnection):
    PipeConnection.send(20)
    PipeConnection.close()

# Per la comunicazione tra processi, devo usare una pipe che gestisce sia lo scambio
# che la cooperazione (produttore e consumatore automatico)
conn1, conn2 = mp.Pipe()

p1 = mp.Process(target=f,args=(conn1,))
p2 = mp.Process(target=g,args=(conn2,))

p1.start()
p2.start()

print(conn1.recv())
print(conn2.recv())

p1.join()
p2.join()

# Aggiungendo duplex = False faccio in modo che ciascuna pipe sia unidirezionale.
# Questo vuol dire che una delle pipe è send-only, l'altra recv-only
SendPipe, RecvPipe = mp.Pipe(duplex=False)

p1 = mp.Process(target=f,args=(SendPipe,))
p2 = mp.Process(target=g,args=(RecvPipe,)) # da errore perchè provo a mandare su una coda recv

p1.start()
p2.start()

print(SendPipe.recv())
print(RecvPipe.recv())

p1.join()
p2.join()
# Alternativamente posso usare una Queue, anche se questa non gestisce automaticamente
# il problema dei produttori e consumatori. Questo vuol dire che devo utilizzare dei semafori
# o dei lock per gestirle manualmente.
import multiprocessing as mp
from time import sleep

def f(QueueObj):
    print("Processo put -> Inserisco 10 nella coda..")
    sleep(3)
    QueueObj.put(10)
    print("Processo put -> Inserito 10 nella coda\n")

def g(QueueObj):
    print("Processo get -> Aspetto dato sulla coda..")
    ret = QueueObj.get()
    print(f"Processo get -> Ho ottenuto {ret} dalla coda")

queue = mp.Queue(1) # L'argomento sono il massimo numero di elementi nella coda

p1 = mp.Process(target=f,args=(queue,)).start()
p2 = mp.Process(target=g,args=(queue,)).start()
```

```python
# Un altra alternativa è l'utilizzo delle shared memory implementate nella 
# libreria mp e che si comportano esattamente come le shared memory C.
# Queste shared memory sono thread/process safe, quindi la sincronizzazione
# è gestita automaticamente.
import multiprocessing as mp

# Le classi Value e Array danno oggetti allocati su una shared memory
# Il primo parametro è il tipo della variabile, il secondo il valore
val = mp.Value("d",1)
arr = mp.Array("i",[1,2,3])

print(f"Valore di val prima della chiamata a funzione: {val.value}")
print(f"Valore di arr prima della chiamata a funzione: {arr[:]}\n")

def f(val,arr):
    val.value = 10

    for i in range(len(arr)):
        arr[i] += 1

p = mp.Process(target=f,args=(val,arr))

p.start()
p.join()

print(f"Valore di val dopo la chiamata a funzione: {val.value}")
print(f"Valore di arr dopo la chiamata a funzione: {arr[:]}")
```

```python
# La keyword yield viene utilizzata per ritornare un valore al chiamante senza terminare la funzione
N=3

def f(n):
    """
    Calcola i quadrati di tutti i numeri da 0 a n-1
    """
    for i in range(n):
        yield i**2

a = f(N)
# Un oggetto iterabile che contiene i valori generati dalla funzione.
print(f"Tipo di a: {type(a)}\n")

for i in range(N):
    # Ad ogni iterazione devo aumentare l'indice interno dell'iteratore
    # per prendere l'elemento successivo
    print(f"Valore n.{i+1} generato dalla funzione: {next(a)}")
```

```python
# Usa https://httpsbin.com per avere un server che risponde con le stesse cose che gli invii
import requests

BASE_URL = "https://httpbin.org"

def print_all(response_obj):
    print(f"Status Code: {response_obj.status_code}")
    print(f"Contenuto: {response_obj.text}")
    print(f"Headers: {response_obj.headers}")

# Head
r_head = requests.head(BASE_URL+"/get")
print_all(r_head)

# Get
r_get = requests.get(BASE_URL+"/get")
print_all(r_get)

# Post
dic = {"key":"pls answer", "cock":"balls"}
r_post = requests.post(BASE_URL+"/post",data=dic)
print_all(r_post)
```
